```python
# ---------------- Imports ----------------
# Standard
import random
import os
import json

# External
import pandas as pd
import yaml


```

```python
# ---------------- Args ----------------
input_file = "combined-claims"
sample_size = 1000
seed = 42
```

```python
# ---------------- Config ----------------

with open("../config/config.yaml", "r") as f:
    config = yaml.safe_load(f)


random.seed(seed)

PROJ_STORE = config["paths"]["proj-store"]
DATA_FOLDER = os.path.join(PROJ_STORE, "data")


INPUT_DIR = os.path.join(DATA_FOLDER, "combined-claims")
input_path = os.path.join(INPUT_DIR, f"{input_file}.jsonl")

# OUTPUT
output_path = os.path.join(INPUT_DIR, f"{input_file}-{sample_size}.jsonl")

```

```python
# ---------------- Functions ----------------

# Read all lines
with open(input_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

# Sample
sampled_lines = random.sample(lines, sample_size)

# Write output
with open(output_path, "w", encoding="utf-8") as f:
    for line in sampled_lines:
        f.write(line)

```
